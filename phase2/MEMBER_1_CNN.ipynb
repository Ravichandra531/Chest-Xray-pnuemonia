## CNN Architecture Overview

```
Input (256x256 RGB)
    ↓
Block 1: Conv→BN→ReLU→Conv→BN→ReLU→MaxPool→Dropout (32 filters)
    ↓
Block 2: Conv→BN→ReLU→Conv→BN→ReLU→MaxPool→Dropout (64 filters)
    ↓
Block 3: Conv→BN→ReLU→Conv→BN→ReLU→MaxPool→Dropout (128 filters)
    ↓
Block 4: Conv→BN→ReLU→Conv→BN→ReLU→MaxPool→Dropout (256 filters)
    ↓
Block 5: Conv→BN→ReLU→Conv→BN→ReLU→MaxPool→Dropout (512 filters)
    ↓
Global Average Pooling (512×1×1)
    ↓
Classifier: FC(512→256→128→2)
    ↓
Output: [class_0_logit, class_1_logit]
```

**Total Parameters**: ~13M
**Target Performance**: ≥85% CV balanced accuracy

## Key Concepts

### 1. Convolutional Layers
- Extract spatial features using learnable filters
- Each filter slides across the image detecting patterns
- Filter size: 3×3 (standard for medical imaging)
- Padding: 1 (preserves spatial dimensions)

### 2. Batch Normalization
- Normalizes layer inputs to stabilize training
- Reduces internal covariate shift
- Allows higher learning rates
- Improves generalization

### 3. ReLU Activation
- Introduces non-linearity: max(0, x)
- Enables the network to learn complex patterns
- Computationally efficient

### 4. Max Pooling
- Downsamples spatial dimensions
- Reduces computation and memory
- Extracts most important features

### 5. Dropout
- Regularization technique to prevent overfitting
- Randomly deactivates neurons during training
- Rate: 0.3 (30% of neurons dropped)

### 6. Global Average Pooling
- Converts feature maps to vector
- Reduces parameters and overfitting
- More robust than flattening

## Viva Preparation Questions

**Q1: Explain the role of batch normalization in your CNN**
A: Batch normalization normalizes layer inputs, stabilizing training and allowing higher learning rates. It reduces internal covariate shift and improves generalization.

**Q2: Why use both conv blocks and dropout?**
A: Conv blocks extract features, dropout prevents overfitting by randomly deactivating neurons. This combination balances feature learning with generalization.

**Q3: What is early stopping and why use it?**
A: Early stopping monitors validation accuracy and stops training when it stops improving (patience=10). This prevents overfitting while maintaining computational efficiency.

**Q4: How does global average pooling improve the model?**
A: Global average pooling reduces parameters, decreases overfitting risk, and is more robust than flattening. It provides spatial translation invariance.

**Q5: Describe your data augmentation strategy**
A: Rotation (±20°), horizontal flip (50%), and brightness adjustment (±20%) on training data. This increases data diversity and improves generalization without overfitting to specific image orientations or lighting.
